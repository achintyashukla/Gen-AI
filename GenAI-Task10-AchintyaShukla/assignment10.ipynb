### Task 1: Pandas Series Basics

```
import pandas as pd


# 2. Create a Pandas Series from the list:
marks = [78, 85, 90, 66, 72]
series_marks = pd.Series(marks)

# 4. Print: Series values, Index, Data type
print("Series values:")
print(series_marks.values)
print("\nIndex:")
print(series_marks.index)
print("\nData type:")
print(series_marks.dtype)

# 5. Access: First element, Last two elements
print("\nFirst element:")
print(series_marks.iloc[0])
print("\nLast two elements:")
print(series_marks.iloc[-2:])
```

### Task 2: Mathematical Operations on Series

```
# Using the marks Series:
# 1. Add 5 grace marks to all students.
print("Add 5 grace marks:")
print(series_marks + 5)

# 2. Subtract 2 marks from all values.
print("\nSubtract 2 marks:")
print(series_marks - 2)

# 3. Multiply all marks by 1.05.
print("\nMultiply all marks by 1.05:")
print(series_marks * 1.05)

# 4. Divide all marks by 2.
print("\nDivide all marks by 2:")
print(series_marks / 2)
```

### Task 3: Python Functionalities on Series

```
# Using the same Series:
# 1. Find: Maximum marks, Minimum marks, Sum of marks, Mean marks
print(f"Maximum marks: {series_marks.max()}")
print(f"Minimum marks: {series_marks.min()}")
print(f"Sum of marks: {series_marks.sum()}")
print(f"Mean marks: {series_marks.mean()}")

# 2. Apply a lambda function to check whether each student has passed (>= 70).
passed_series = series_marks.apply(lambda x: x >= 70)
print("\nPassed (>= 70):")
print(passed_series)

# 3. Count how many students passed.
print(f"\nNumber of students passed: {passed_series.sum()}")
```

### Task 4: Create a DataFrame

```
students = {
    'Name': ['Amit', 'Neha', 'Rahul', 'Sneha', 'Pooja'],
    'Marks': [78, 85, 90, 66, 72],
    'Subject': ['Math', 'Math', 'Science', 'Science', 'Math']
}
# 1. Convert it into a DataFrame.
df_students = pd.DataFrame(students)

# 2. Print first 3 rows.
print("First 3 rows:")
print(df_students.head(3))

# 3. Print last 2 rows.
print("\nLast 2 rows:")
print(df_students.tail(2))

# 4. Print DataFrame shape and column names.
print(f"\nDataFrame shape: {df_students.shape}")
print(f"Column names: {list(df_students.columns)}")
```

### Task 5: Important DataFrame Functions

```
# 1. Use and print output of: .info(), .describe(), .head(), .tail()
print("Info:")
df_students.info()

print("\nDescribe:")
print(df_students.describe())

print("\nHead:")
print(df_students.head())

print("\nTail:")
print(df_students.tail())

# 2. Sort students by Marks in descending order.
df_sorted = df_students.sort_values(by='Marks', ascending=False)
print("\nSorted by Marks (descending):")
print(df_sorted)

# 3. Reset index after sorting.
df_sorted_reset = df_sorted.reset_index(drop=True)
print("\nReset index after sorting:")
print(df_sorted_reset)
```

### Task 6: Filtering & Conditional Selection

```
# 1. Students who scored more than 75 marks.
print("Students who scored more than 75 marks:")
print(df_students[df_students['Marks'] > 75])

# 2. Students belonging to subject Math.
print("\nStudents belonging to subject Math:")
print(df_students[df_students['Subject'] == 'Math'])

# 3. Students who scored more than average marks.
avg_marks = df_students['Marks'].mean()
print(f"\nStudents who scored more than average marks ({avg_marks}):")
print(df_students[df_students['Marks'] > avg_marks])

# 4. Students who failed (marks < 70).
print("\nStudents who failed (marks < 70):")
print(df_students[df_students['Marks'] < 70])
```

### Task 7: Grouping & Basic Analysis

```
# 1. Find average marks per subject using groupby().
print("Average marks per subject:")
print(df_students.groupby('Subject')['Marks'].mean())

# 2. Count number of students per subject.
print("\nNumber of students per subject:")
print(df_students.groupby('Subject')['Name'].count())

# 3. Find maximum marks per subject.
print("\nMaximum marks per subject:")
print(df_students.groupby('Subject')['Marks'].max())
```

### Task 8: Pandas Plotting (Simple Graphs)

```
# 1. Plot a bar graph of student names vs marks.
print("Creating bar graph (Name vs Marks)...")
df_students.plot(kind='bar', x='Name', y='Marks', title='Student Names vs Marks')

# 2. Plot a line graph of marks.
print("Creating line graph of marks...")
df_students['Marks'].plot(kind='line', title='Line Graph of Marks')

# 3. Plot a histogram of marks.
print("Creating histogram of marks...")
df_students['Marks'].plot(kind='hist', title='Histogram of Marks')
print("(Plots created. In a Jupyter notebook or with matplotlib.pyplot.show(), they would be displayed.)")
```

### Task 9: Mini Use Case: Sales Data Analysis

```
sales = {
    'Day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri'],
    'Revenue': [1200, 1500, 900, 2000, 1800]
}
df_sales = pd.DataFrame(sales)

# 1. Total revenue
total_revenue = df_sales['Revenue'].sum()
print(f"Total revenue: {total_revenue}")

# 2. Average daily revenue
avg_revenue = df_sales['Revenue'].mean()
print(f"Average daily revenue: {avg_revenue}")

# 3. Day with highest revenue
max_revenue_day = df_sales.loc[df_sales['Revenue'].idxmax(), 'Day']
print(f"Day with highest revenue: {max_revenue_day}")

# 4. Days where revenue > average
print("\nDays where revenue > average:")
print(df_sales[df_sales['Revenue'] > avg_revenue])

# 5. Plot revenue vs day
print("\nCreating plot of revenue vs day...")
df_sales.plot(kind='line', x='Day', y='Revenue', title='Revenue vs Day', marker='o')
print("(Plot created.)")
```
