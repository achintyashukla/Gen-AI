### Task 1: Write Sales Records to a File


```
sales = [1200, 450, 980, 1500, 3000]

# Write data to file
with open('sales_data.txt', 'w') as f:
    for sale in sales:
        f.write(str(sale) + '\n')

# Close is automatically handled by 'with' block, now reopen and print
print("Contents of sales_data.txt:")
with open('sales_data.txt', 'r') as f:
    print(f.read())

# Extra: comma-separated format
with open('sales_data_csv.txt', 'w') as f:
    f.write(",".join(map(str, sales)) + '\n')

```

### Task 2: Read File in Different Ways


```
print("--- 1. Using .read() ---")
with open('sales_data.txt', 'r') as f:
    content = f.read()
    print(content)

print("--- 2. Using .readline() ---")
with open('sales_data.txt', 'r') as f:
    first_line = f.readline()
    print("First line:", first_line.strip())

print("\n--- 3. Using .readlines() to list of integers ---")
with open('sales_data.txt', 'r') as f:
    lines = f.readlines()
    
# convert to int, strip newlines
sales_list = []
for line in lines:
    sales_list.append(int(line.strip()))
    
print("List of integers:", sales_list)

```

### Task 3: Append New Sales


```
new_sales = [5000, 2500, 1700]

# Append new sales
with open('sales_data.txt', 'a') as f:
    for sale in new_sales:
        f.write(str(sale) + '\n')

print("Updated contents of sales_data.txt:")
with open('sales_data.txt', 'r') as f:
    updated_content = f.read()
    print(updated_content)

# Extra: Print total number of lines
with open('sales_data.txt', 'r') as f:
    total_lines = len(f.readlines())
    print("Total number of lines after appending:", total_lines)

```

### Task 4: Generate Summary Report from File


```
with open('sales_data.txt', 'r') as f:
    lines = f.readlines()

# Convert to integers
all_sales = []
for line in lines:
    all_sales.append(int(line.strip()))

total_sales = sum(all_sales)
highest_sale = max(all_sales)
lowest_sale = min(all_sales)
average_sale = total_sales / len(all_sales) if len(all_sales) > 0 else 0

print("Summary Report:")
print(f"Total Sales: {total_sales}")
print(f"Highest Sale: {highest_sale}")
print(f"Lowest Sale: {lowest_sale}")
print(f"Average Sale: {average_sale:.2f}")

```

### Task 5: Create Product Info File (User Input)


```
# Note: In a real environment, you'd use input() here.
# For example: 
# prod_name = input("Enter Product Name: ")
# prod_price = input("Enter Product Price: ")

# We use hardcoded inputs below so the notebook can execute safely without hanging
products_info = [
    ("Laptop", "1000"),
    ("Mouse", "50"),
    ("Keyboard", "80")
]

with open('products.txt', 'w') as f:
    for name, price in products_info:
        f.write(f"{name} | {price}\n")

print("Reading from products.txt:")
with open('products.txt', 'r') as f:
    for line in f:
        print(line.strip())

```

### Task 6: Read File Safely (Error Handling Inside File Handling Only)


```
import os

# filename = input("Enter a filename to open: ")
# We use a hardcoded filename to prevent blocking execution
filename = "sales_data.txt" 

if os.path.exists(filename):
    print(f"--- Contents of {filename} ---")
    with open(filename, 'r') as f:
        print(f.read())
else:
    print("File not found. Please check the filename.")
    
# Testing with a missing file
missing_file = "does_not_exist.txt"
if os.path.exists(missing_file):
    with open(missing_file, 'r') as f:
        print(f.read())
else:
    print(f"\nTrying to open '{missing_file}':")
    print("File not found. Please check the filename.")

```

### Task 7: Mini Project - Export Discounted Prices


```
prices = {
    "Mouse": 500,
    "Keyboard": 800,
    "Monitor": 7000,
    "Pendrive": 400,
    "Camera": 5000
}

# discount_pct_str = input("Enter discount percentage: ")
# Hardcoded to 15 for demonstration
discount_pct_str = "15" 
discount_pct = float(discount_pct_str)

total_items = len(prices)
sum_discounted = 0

with open('discount_report.txt', 'w') as f:
    f.write("Product | Original Price | Discounted Price\n")
    
    for product, orig_price in prices.items():
        disc_price = orig_price - (orig_price * (discount_pct / 100))
        sum_discounted += disc_price
        
        f.write(f"{product} | {orig_price} | {disc_price:.2f}\n")
        
    avg_disc_price = sum_discounted / total_items if total_items > 0 else 0
    
    # Extra: summary at bottom
    f.write("\n--- Summary ---\n")
    f.write(f"Total Items: {total_items}\n")
    f.write(f"Average Discounted Price: {avg_disc_price:.2f}\n")

print("discount_report.txt has been generated!\n")
print("Reading file contents back to terminal:\n")

with open('discount_report.txt', 'r') as f:
    print(f.read())

```
