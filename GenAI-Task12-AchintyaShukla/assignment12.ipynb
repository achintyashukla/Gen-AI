# Assignment 12: Seaborn (Relational, Distribution, Categorical & Multi-Plots)

Dataset used: Kaggle Medical Cost Personal Dataset (`insurance.csv`).  
Libraries used: `pandas`, `matplotlib.pyplot`, and `seaborn`.

### Dataset Loading (Pandas)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("insurance.csv")
print("Dataset Shape:", df.shape)
print(df.head())
```

### Task 1: Relational Plot

```python
# 1 & 2. Relational plot with hue using a categorical column
g1 = sns.relplot(data=df, x="bmi", y="charges", hue="smoker")
g1.fig.suptitle("BMI vs Charges by Smoker Status", y=1.02)
plt.show()

# 3. Same plot using scatter style
g2 = sns.relplot(data=df, x="bmi", y="charges", hue="smoker", style="smoker", kind="scatter")
g2.fig.suptitle("BMI vs Charges (Scatter Style)", y=1.02)
plt.show()
```

### Task 2: Line Plot as Scatter & Facet

```python
# 1. Line plot using sns.lineplot()
sns.lineplot(data=df, x="age", y="charges")
plt.title("Average Charges by Age (Line Plot)")
plt.show()

# 2. Scatter-style line plot (markers without connecting line)
sns.lineplot(data=df, x="age", y="charges", marker="o", linestyle="")
plt.title("Average Charges by Age (Scatter-Style Line Plot)")
plt.show()

# 3. Faceted line plot based on a categorical column
g = sns.relplot(data=df, x="age", y="charges", kind="line", col="smoker")
g.fig.suptitle("Age vs Charges Faceted by Smoker", y=1.03)
plt.show()
```

### Task 3: Distribution Plots

```python
# 1. Histogram
sns.histplot(data=df, x="charges", bins=30)
plt.title("Distribution of Insurance Charges (Histogram)")
plt.show()

# 2. KDE plot
sns.kdeplot(data=df, x="charges")
plt.title("Distribution of Insurance Charges (KDE Plot)")
plt.show()

# 3. Rug plot
sns.rugplot(data=df, x="charges")
plt.title("Distribution of Insurance Charges (Rug Plot)")
plt.show()

# 4. Combined Histogram + KDE
sns.histplot(data=df, x="charges", bins=30, kde=True)
plt.title("Distribution of Insurance Charges (Histogram + KDE)")
plt.show()
```

### Task 4: Bivariate Distribution Plots

```python
# 1. Bivariate histogram
sns.histplot(data=df, x="age", y="bmi")
plt.title("Bivariate Histogram: Age vs BMI")
plt.show()

# 2. Bivariate KDE plot
sns.kdeplot(data=df, x="age", y="bmi", fill=True)
plt.title("Bivariate KDE Plot: Age vs BMI")
plt.show()
```

### Task 5: Matrix Plots

```python
# 1. Pair plot
g = sns.pairplot(df, hue="smoker")
g.fig.suptitle("Pair Plot of Insurance Dataset", y=1.02)
plt.show()

# 2. Correlation matrix heatmap
corr_matrix = df.select_dtypes(include="number").corr()
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix Heatmap")
plt.show()
```

### Task 6: Categorical Plots

```python
# 1. Bar plot
sns.barplot(data=df, x="region", y="charges")
plt.title("Average Charges by Region (Bar Plot)")
plt.show()

# 2. Box plot
sns.boxplot(data=df, x="region", y="charges")
plt.title("Charges Distribution by Region (Box Plot)")
plt.show()

# 3. Violin plot
sns.violinplot(data=df, x="region", y="charges")
plt.title("Charges Distribution by Region (Violin Plot)")
plt.show()

# 4. Count plot
sns.countplot(data=df, x="region")
plt.title("Number of Records per Region (Count Plot)")
plt.show()
```

### Task 7: Regression Plots

```python
# 1. Regression plot (regplot)
sns.regplot(data=df, x="age", y="charges")
plt.title("Regression Plot: Age vs Charges")
plt.show()

# 2. lmplot with hue using a categorical column
g = sns.lmplot(data=df, x="bmi", y="charges", hue="smoker")
g.fig.suptitle("LM Plot: BMI vs Charges by Smoker", y=1.02)
plt.show()
```

### Task 8: Multi-Plots & Figure-Level Plots

```python
# 1. FacetGrid with two numerical variables and a categorical column
grid = sns.FacetGrid(df, col="smoker")
grid.map_dataframe(sns.scatterplot, x="bmi", y="charges")
grid.fig.suptitle("FacetGrid: BMI vs Charges by Smoker", y=1.03)
plt.show()

# 2. Multi-plot dashboard using figure-level functions: relplot, catplot, displot
g_rel = sns.relplot(data=df, x="age", y="charges", hue="smoker", col="sex", kind="scatter")
g_rel.fig.suptitle("Figure-Level relplot: Age vs Charges by Sex & Smoker", y=1.03)
plt.show()

g_cat = sns.catplot(data=df, x="region", y="charges", hue="smoker", kind="box")
g_cat.fig.suptitle("Figure-Level catplot: Charges by Region & Smoker", y=1.03)
plt.show()

g_dis = sns.displot(data=df, x="charges", col="smoker", kde=True)
g_dis.fig.suptitle("Figure-Level displot: Charges Distribution by Smoker", y=1.03)
plt.show()
```
